**Física Mecánica | Dinámica – Leyes de Newton**

---

## 🧠 Conceptos Fundamentales

### 🔹 Masa
La **masa ($m$)** es la cantidad de materia que posee un cuerpo.  
- Unidad en el S.I.: **kg**  
- Es una **magnitud escalar**  
- No debe confundirse con el **peso**

---

### 🔹 Peso
El **peso ($P$)** es la fuerza con que la Tierra atrae un cuerpo.  

$$
P = m \cdot g
$$

Donde:  
- $m$: masa del objeto (kg)  
- $g$: aceleración de la gravedad (9.81 m/s²)  

---

### 🔹 Fuerza
La **fuerza ($F$)** es toda interacción capaz de modificar el estado de movimiento de un cuerpo.  
- Es una **magnitud vectorial**
- Unidad: **Newton (N = kg·m/s²)**

---

## ⚖️ Fuerza Neta y Equilibrio

La **fuerza neta ($\Sigma F$)** es la suma vectorial de todas las fuerzas que actúan sobre un cuerpo.

$$
\Sigma \vec{F} = m \vec{a}
$$

Cuando un cuerpo está en equilibrio:

$$
\Sigma \vec{F} = 0
$$

---

## 🧩 Tipos de Fuerzas

| Tipo de Fuerza | Descripción | Fórmula o Notación |
|-----------------|--------------|--------------------|
| **Peso (P)** | Fuerza debida a la gravedad | $P = m g$ |
| **Normal (N)** | Perpendicular a la superficie de apoyo | $N = m g \cos(\theta)$ |
| **Fricción (Ff)** | Oposición al movimiento por contacto | $F_f = \mu N$ |
| **Tensión (T)** | Fuerza transmitida por una cuerda | $T = m g$ (en equilibrio) |
| **Elástica (Fe)** | Fuerza ejercida por un resorte | $F_e = -k \Delta x$ |

---

## 🧭 Descomposición del Peso en un Plano Inclinado

En un plano inclinado, el peso se descompone en dos componentes:

$$
\begin{cases}
P_{\parallel} = m g \sin(\theta) \\
P_{\perp} = m g \cos(\theta)
\end{cases}
$$

Donde:  
- $P_{\parallel}$ → componente paralela al plano (causa el deslizamiento)  
- $P_{\perp}$ → componente perpendicular (equivale a la fuerza normal $N$)

---

## 🧲 Fuerza de Rozamiento

La **fuerza de fricción ($F_f$)** se opone al movimiento relativo entre superficies en contacto.  
Existen dos tipos:

1. **Fricción estática ($F_e$):**
   - Se presenta cuando no hay movimiento.
   - Aumenta con la fuerza aplicada hasta un máximo $F_{e,\text{max}}$.
   - Fórmula:

     $$
     F_{e,\text{max}} = \mu_e N
     $$

2. **Fricción dinámica ($F_d$):**
   - Actúa cuando el cuerpo ya está en movimiento.
   - Tiene un valor constante:

     $$
     F_d = \mu_d N
     $$

   - Con $\mu_d < \mu_e$

---

## ⚗️ Coeficiente de Fricción Estático

El **coeficiente de fricción estático ($\mu_e$)** se puede determinar experimentalmente mediante el ángulo crítico de inclinación del plano ($\theta_c$):

$$
\mu_e = \tan(\theta_c)
$$

Si el bloque comienza a deslizarse a un ángulo de $25°$:

$$
\mu_e = \tan(25°) = 0.466
$$

---

## 🔬 Ejemplo Teórico

Un bloque de masa **1 kg** se encuentra sobre un plano inclinado de **25°**.  
Calcular las fuerzas que actúan sobre él:

$$
\begin{align}
P &= m g = 1(9.81) = 9.81 \, N \\
P_{\parallel} &= 9.81 \sin(25°) = 4.15 \, N \\
P_{\perp} &= 9.81 \cos(25°) = 8.89 \, N \\
\mu_e &= \tan(25°) = 0.466 \\
F_f &= \mu_e N = 0.466(8.89) = 4.14 \, N
\end{align}
$$

✅ **Resultado:** La fuerza de fricción equilibra la componente paralela, impidiendo el movimiento.  
En el instante del deslizamiento:

$$
F_f = P_{\parallel}
$$

## ⚙️ Leyes de Newton

### 🥇 Primera Ley – Ley de la Inercia
> Un cuerpo permanecerá en reposo o en movimiento uniforme si la fuerza neta sobre él es cero.  

$$
\Sigma \vec{F} = 0
$$

---

### 🥈 Segunda Ley – Ley Fundamental de la Dinámica
> La fuerza neta aplicada sobre un cuerpo es igual al producto de su masa por su aceleración.  

$$
\Sigma \vec{F} = m \vec{a}
$$

---

### 🥉 Tercera Ley – Acción y Reacción
> Toda acción tiene una reacción igual y opuesta.  

$$
\vec{F}_{12} = -\vec{F}_{21}
$$

---

## 🧮 Relación entre masa, área y fricción

- El **coeficiente de fricción ($\mu$)** **no depende**:
  - del área de contacto  
  - de la masa del objeto  

- Sin embargo, la **fuerza de fricción ($F_f$)** sí depende de la masa a través de:  

  $$
  N = m \cdot g \cdot \cos(\theta)
  $$
