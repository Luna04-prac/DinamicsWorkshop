# **Ejercicio 10 - Sistema sin rozamiento**

Conjunto de bloques unidos por cuerdas y poleas sin fricción.
Se pide calcular la aceleración y las tensiones.

```python
from sympy import symbols, simplify

# Declaramos las variables simbólicas
# m1, m2, m3: masas del sistema
# g: gravedad
# a: aceleración del conjunto
# T1, T2: tensiones en las cuerdas
m1, m2, m3, g, a, T1, T2 = symbols('m1 m2 m3 g a T1 T2')

# La fuerza neta sobre el sistema proviene del bloque colgante m3:
# m3*g = (m1 + m2 + m3)*a
a_expr = simplify((m3*g) / (m1 + m2 + m3))

# Tensiones internas en las cuerdas:
# T1 = m1*a   (tensión entre m1 y m2)
# T2 = (m1 + m2)*a   (tensión entre m2 y m3)
T1_expr = simplify(m1*a_expr)
T2_expr = simplify((m1 + m2)*a_expr)

# Resultados generales
print("=== Sistema sin rozamiento (Análisis simbólico) ===")
print(f"Aceleración general: a = {a_expr}")
print(f"Tensión entre m1 y m2: T1 = {T1_expr}")
print(f"Tensión entre m2 y m3: T2 = {T2_expr}")
print("→ Estas expresiones representan las relaciones generales entre las masas.")
```
