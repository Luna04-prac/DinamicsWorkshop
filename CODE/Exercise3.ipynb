# **Ejercicio 3 - Fuerza mínima para mover m1**

Sistema de dos bloques m1 y m2. El coeficiente de fricción entre
m1 y la mesa es μ1, y entre m1 y m2 es μ2.
Se desea calcular la fuerza F mínima para sacar a m1 del reposo.

`Datos de ejemplo:`
m1 = 10 kg
m2 = 4 kg
μ1 = 0.3
μ2 = 0.2

```python
from sympy import symbols, simplify

# Declaración de variables simbólicas
# μ1, μ2: coeficientes de fricción
# m1, m2: masas
# g: gravedad
# F: fuerza aplicada
m1, m2, μ1, μ2, g, F = symbols('m1 m2 μ1 μ2 g F')

# La fuerza mínima para iniciar el movimiento debe igualar
# la suma de las fuerzas de fricción en cada contacto.
F_min = simplify((μ1*m1 + μ2*m2) * g)

# Mostramos la ecuación final
print("=== Fuerza mínima para mover m1 (Análisis simbólico) ===")
print(f"F_min = {F_min}")
print("→ La fuerza debe superar la fricción combinada de ambas superficies.")
```
