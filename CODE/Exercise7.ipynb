# **Ejercicio 7 - Fuerza inclinada con polea**

Un bloque de masa m sobre superficie con rozamiento (μ)
es arrastrado por una fuerza F inclinada un ángulo α y unido
por cuerda a una masa colgante M.

Se pide:
a) Aceleración del sistema
b) Tensión en la cuerda

```python
from sympy import symbols, sin, cos, simplify

# Declaración de variables simbólicas
m, M, μ, g, a, T, α = symbols('m M μ g a T α')

# Consideramos las ecuaciones dinámicas simplificadas:
# Masa colgante:   M*g - T = M*a
# Bloque horizontal: T - μ*m*g = m*a

# Despejamos a en términos de las variables conocidas:
a_expr = simplify((M*g - μ*m*g) / (m + M))

# Calculamos la tensión reemplazando a en la ecuación del bloque colgante:
T_expr = simplify(M*(g - a_expr))

# Mostramos los resultados
print("=== Sistema con fuerza inclinada (Análisis simbólico) ===")
print(f"Aceleración general: a = {a_expr}")
print(f"Tensión general: T = {T_expr}")
print("→ El resultado muestra la dependencia entre masas y fricción, sin números concretos.")
```
