## 📚 Tema: Aplicaciones de las Leyes de Newton

### Nota importante:
> La aceleración es la misma para todo el sistema dado que existe una **polea única**.

---

## ⚙️ Análisis del Sistema

El sistema está compuesto por dos masas **m₁** y **m₂** conectadas por una cuerda ideal que pasa por una polea sin fricción.  
Cada masa se encuentra sometida a distintas fuerzas según su posición (plano, superficie o suspensión).

---

### 🔹 Fuerzas que actúan sobre el sistema:

- **N₁, N₂:** Fuerzas normales (perpendiculares a la superficie).  
- **P₁, P₂:** Pesos de cada cuerpo, dados por *P = m·g*.  
- **f₁, f₂:** Fuerzas de fricción (si existen).  
- **T:** Tensión de la cuerda.  

---

## 🔩 Diagramas de Cuerpo Libre (DCL)

Los diagramas de cuerpo libre permiten visualizar las fuerzas que actúan sobre cada bloque del sistema.

### Bloque 1:
- Peso: P₁ = m₁·g  
- Normal: N₁  
- Fricción: f₁ (si aplica)  
- Tensión: T  
- Aceleración: a  

### Bloque 2:
- Peso: P₂ = m₂·g  
- Normal: N₂  
- Fricción: f₂ (si aplica)  
- Tensión: T  
- Aceleración: a  

---

## ⚖️ Segunda Ley de Newton (aplicada al sistema)

Para cada masa:

**Bloque 1 (en superficie horizontal o inclinada):**
> ΣF₁ = T - f₁ - (componente del peso si hay inclinación) = m₁·a

**Bloque 2 (colgante):**
> ΣF₂ = P₂ - T = m₂·a

---

### 🧩 Si hay fricción:
> f = μ·N  
> N = m·g·cos(θ)

---

## 🧮 Ecuación del sistema (sin fricción)

Cuando la cuerda y la polea son ideales:

> a = (m₂·g) / (m₁ + m₂)

Y la tensión se obtiene como:

> T = m₁·a + f₁  → si hay fricción  
> T = m₁·a  → si no hay fricción

---

## 🔬 Ejemplo numérico

Supón que:
- m₁ = 2 kg (sobre superficie lisa)  
- m₂ = 3 kg (colgando)  
- g = 9.8 m/s²  

**Cálculo:**

> a = (m₂·g) / (m₁ + m₂)  
> a = (3 × 9.8) / (2 + 3)  
> a = 5.88 m/s²  

> T = m₁·a = 2 × 5.88 = 11.76 N  

**Resultados:**  
- Aceleración del sistema: **5.88 m/s²**  
- Tensión en la cuerda: **11.76 N**

---

## ⚗️ Caso con fricción en el plano

Si el bloque 1 está sobre una superficie con fricción (coeficiente μ):

> f₁ = μ·N = μ·m₁·g  

Entonces:

> a = (m₂·g - f₁) / (m₁ + m₂)

**Ejemplo:**
- μ = 0.2  
- m₁ = 2 kg, m₂ = 3 kg  

> f₁ = 0.2 × 2 × 9.8 = 3.92 N  
> a = (3×9.8 - 3.92) / (2 + 3) = 4.54 m/s²  
> T = m₁·a + f₁ = 2×4.54 + 3.92 = 12.99 N  

**Resultados con fricción:**  
- Aceleración: **4.54 m/s²**  
- Tensión: **12.99 N**

---

## 🧠 Conclusiones

1. La aceleración del sistema depende de la diferencia de fuerzas entre las masas conectadas.  
2. La fricción disminuye la aceleración y aumenta la tensión.  
3. En un sistema con polea ideal, ambas masas comparten la misma magnitud de aceleración.  
4. Las Leyes de Newton permiten analizar cualquier configuración de fuerzas aplicadas sobre cuerpos en movimiento o equilibrio.
