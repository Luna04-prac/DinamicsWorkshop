# **Ejercicio 1 - Máquina de Atwood**

Sistema con dos masas m1 y m2 unidas por una cuerda ideal
que pasa sobre una polea sin fricción. Se pide:
a) Describir el movimiento.
b) Calcular la aceleración de cada masa.
c) Determinar la tensión en la cuerda.

`Datos de ejemplo:`
- m1 = 5 kg
- m2 = 3 kg
- g = 9.81 m/s²

```python
from sympy import symbols, Eq, solve, simplify

# Declaramos las variables simbólicas
# m1, m2: masas
# g: gravedad
# a: aceleración del sistema
# T: tensión en la cuerda
m1, m2, g, a, T = symbols('m1 m2 g a T')

# Planteamos las ecuaciones de Newton para cada masa
# Para m1 (que desciende):  m1*g - T = m1*a
# Para m2 (que asciende):   T - m2*g = m2*a
eq1 = Eq(m1*g - T, m1*a)
eq2 = Eq(T - m2*g, m2*a)

# Resolvemos el sistema de ecuaciones para a y T
sol = solve((eq1, eq2), (a, T))

# Mostramos los resultados de manera simbólica
print("=== Máquina de Atwood (Análisis simbólico) ===")
print("Aceleración general del sistema:")
print(f"a = {simplify(sol[a])}")   # Expresión simplificada para a

print("\nTensión general en la cuerda:")
print(f"T = {simplify(sol[T])}")   # Expresión simplificada para T
```
