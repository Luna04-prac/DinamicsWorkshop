# **Ejercicio 2 - Ascensor con contrapeso**

El ascensor (m1 = 1100 kg) y el contrapeso (m2 = 1000 kg)
están unidos por un cable y polea ideal. El ascensor acelera
hacia arriba a 2 m/s².

`Calcular:`
a) Tensión T1 en el cable que sostiene la caja
b) Tensión T2 en el tramo del contrapeso


```python
# Datos
m1 = 1100  # kg
m2 = 1000  # kg
a = 2      # m/s²
g = 9.81   # m/s²

# Cálculos
T1 = m1 * (g + a)
T2 = m2 * (g - a)

# Resultados
print("=== Ascensor con contrapeso ===")
print(f"Tensión en el cable del ascensor (T1): {T1:.2f} N")
print(f"Tensión en el cable del contrapeso (T2): {T2:.2f} N")
```
