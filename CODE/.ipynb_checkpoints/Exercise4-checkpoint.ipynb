# **Ejercicio 4 - Fuerza F al umbral de movimiento**

Sistema de dos masas m1 y m2 con fricción μ1 (mesa) y μ2 (entre bloques).
Se pide la fuerza mínima F para que m1 empiece a moverse.

```python
from sympy import symbols, simplify

# Declaración de variables
m1, m2, μ1, μ2, g = symbols('m1 m2 μ1 μ2 g')

# Cuando el sistema está justo en el umbral del movimiento,
# la fuerza aplicada F es igual a la suma de las fuerzas de fricción.
F_min = simplify((μ1*m1 + μ2*m2) * g)

# Mostramos la ecuación simbólica
print("=== Fuerza al umbral de movimiento (Análisis simbólico) ===")
print(f"F_min = {F_min}")
print("→ La fuerza mínima debe igualar la fricción total antes de iniciar el deslizamiento.")
```
