# **Ejercicio 6 - Velocidad después de 4 s**

Bloques m1 = 20 kg y m2 = 60 kg unidos por cuerda ideal.
Rozamiento cinético μk = 0.3. Hallar la velocidad del sistema
tras 4 s desde el reposo.

```python
# Datos
m1 = 20
m2 = 60
μk = 0.3
t = 4
g = 9.81

# Cálculos
a = (m2 * g - μk * m1 * g) / (m1 + m2)
v = a * t

# Resultados
print("=== Velocidad del sistema ===")
print(f"Aceleración: {a:.2f} m/s²")
print(f"Velocidad después de 4 s: {v:.2f} m/s")
```
