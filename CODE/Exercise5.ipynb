# **Ejercicio 5 - Bloques con rozamiento sobre plano**

Dos bloques m1 y m2 unidos por una cuerda pasan por una polea.
El bloque m0 cuelga. Se desea:
a) Calcular la aceleración del bloque colgante.
b) Calcular la tensión entre m1 y m2.

Datos de ejemplo:
m0 = 4 kg
m1 = 3 kg
m2 = 2 kg
μ = 0.2

```python

from sympy import symbols, Eq, simplify

# Declaración de variables
# m0: masa colgante
# m1, m2: masas sobre la superficie
# μ: coeficiente de fricción
# g: gravedad
# a: aceleración del sistema
# T: tensión entre m1 y m2
m0, m1, m2, μ, g, a, T = symbols('m0 m1 m2 μ g a T')

# Aplicamos la Segunda Ley de Newton al sistema completo
# La fuerza neta que impulsa al sistema es:
#   F = m0*g - Fuerza de fricción total
# donde la fricción total = μ*(m1 + m2)*g

# Aceleración del sistema:
a_expr = simplify((m0*g - μ*(m1 + m2)*g) / (m0 + m1 + m2))

# Tensión entre m1 y m2, considerando las fuerzas horizontales en m1:
T_expr = simplify(m1*(μ*g + a_expr))

# Mostramos resultados generales
print("=== Sistema de bloques con rozamiento (Análisis simbólico) ===")
print(f"Aceleración general: a = {a_expr}")
print(f"Tensión entre m1 y m2: T = {T_expr}")
print("→ Estas expresiones permiten calcular la dinámica sin necesidad de valores numéricos.")
```
