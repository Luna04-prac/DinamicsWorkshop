# **Ejercicio 8 - Bloque con dos cuerdas**

El peso del bloque w es de 80 N. Se desea hallar las
tensiones en las cuerdas A y B que lo sostienen.

```python
from sympy import symbols, sin, Eq, solve, simplify

# Declaración de variables
w, T, θ = symbols('w T θ')

# Condición de equilibrio vertical:
# La suma de las componentes verticales de ambas tensiones
# debe equilibrar el peso del bloque.
eq = Eq(2*T*sin(θ), w)

# Despejamos T simbólicamente
T_expr = simplify(solve(eq, T)[0])

# Mostramos resultado general
print("=== Tensiones en cuerdas A y B (Análisis simbólico) ===")
print(f"Tensión en cada cuerda: T = {T_expr}")
print("→ Cada cuerda soporta la mitad del peso ajustada por el ángulo θ.")
```
